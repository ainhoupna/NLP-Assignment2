## Introduction
In this assignment, we implement and compare two neural architectures for misogyny detection on the MAMI dataset:
1. **BiLSTM** with FastText embeddings.
2. **BERT** (Fine-tuning pre-trained transformers).

We also perform learning curve analysis, ablation studies, and error analysis.

### Setup & Imports

```python
import os
import sys
import json
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from IPython.display import Image, display

# Add src to path
sys.path.append(os.path.abspath(os.path.join('../src')))
sys.path.append(os.path.abspath(os.path.join('..')))

# Constants
RESULTS_DIR = '../results'

def load_json(filename):
    path = os.path.join(RESULTS_DIR, filename)
    if os.path.exists(path):
        with open(path, 'r') as f:
            return json.load(f)
    return None

print("Setup complete.")
```

## Experiment 1: Architecture Comparison
Comparison of three models:
1. **A1 Baseline**: Logistic Regression + FastText (from Assignment 1)
2. **BiLSTM**: Bidirectional LSTM with dropout and frozen FastText embeddings.
3. **BERT**: Fine-tuned BERT-base-uncased with gradual unfreezing.

## Experiment 4: Error Analysis
Comparison of errors between the Baseline and Neural Models.
- **Fixed**: Errors made by Baseline but correctly classified by neural models.
- **New Errors**: Correctly classified by Baseline but missed by neural models.

```python
display(Image(filename=os.path.join(RESULTS_DIR, 'exp4_confusion_matrices.png')))

res4 = load_json('exp4_results.json')
if res4:
    print(f"Baseline Errors: {res4['baseline_errors']}")
    print(f"Fixed by both neural models: {res4['fixed_by_both']}")
    print(f"New errors introduced by both: {res4['new_errors_both']}")
    
    print("\n--- Example Fixed by Neural Models ---")
    ex = res4['fixed_examples'][0]
    print(f"Text: {ex['text']}")
    print(f"True: {ex['true']}, Baseline: {ex['baseline']}, BERT: {ex['bert']}")
```
